###  Check numerical columns for potential outliers

```python
# Check numerical columns for potential outliers

numerical_columns = [
    "Age",
    "Quantity",
    "Unit_Price",
    "Total_Sales"
]

for column in numerical_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ][column].count()
    
    print(f"{column}:")
    print(f"  Lower bound: {lower_bound:.2f}")
    print(f"  Upper bound: {upper_bound:.2f}")
    print(f"  Potential outliers: {outliers}")
    print()
```

### Outlier Assessment

The IQR method identified 19 potential outliers in the `Total_Sales` column. No potential outliers were identified in Age, Quantity, or Unit_Price. The identified Total_Sales values were retained because they are consistent with the relationship `Quantity × Unit_Price` and there was no evidence that they represented data-entry errors.
